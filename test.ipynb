```python
from sklearn.datasets import fetch_openml
import scaleMAP

import matplotlib.pyplot as plt

X = {}
y = {}

# Load MNIST data
mnist = fetch_openml('mnist_784', version=1)
X['mnist'] = mnist.data #[:10000]  # Use a subset for speed
y['mnist'] = mnist.target #[:10000]
```

```python
scaleMAP.__version__
```

```python
umap = scaleMAP.UMAP(verbose=True, n_epochs=800)
embedding_umap = umap.fit_transform(X['mnist'])
```

```python
plt.figure(figsize=(10, 8))
plt.scatter(
    embedding_umap[:, 0],
    embedding_umap[:, 1],
    c=y['mnist'].astype(int),
    cmap='tab10',
    s=1,
)
plt.gca().set_aspect('equal', 'datalim')
plt.axis('off')
plt.tight_layout()
# plt.savefig('umap_mnist.png', dpi=300, bbox_inches='tight')
# plt.close()
```

```python
smap = scaleMAP.UMAP(verbose=True, scalemap=True)
embedding_smap = smap.fit_transform(X['mnist'])
```

```python
plt.figure(figsize=(10, 8))
plt.scatter(
    embedding_smap[:, 0],
    embedding_smap[:, 1],
    c=y['mnist'].astype(int),
    cmap='tab10',
    s=1,
)
plt.gca().set_aspect('equal', 'datalim')
plt.axis('off')
plt.tight_layout()
# plt.savefig('umap_mnist.png', dpi=300, bbox_inches='tight')
# plt.close()
```
